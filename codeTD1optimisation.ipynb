```python

import numpy as np
import matplotlib.pyplot as plt

def df2(x, y):
    # Calcule et renvoie le gradient de f2 au point (x, y)
    df_dx = 3*x**2 - 12*y
    df_dy = -12*x + 24*y**2
    
    return [df_dx, df_dy]


# Fonction qui calcule la matrice hessienne
def d2f2(x, y):
    d2f_dxdx = 6*x
    d2f_dydy = 48*y
    d2f_dxdy = -12
    d2f_dydx = d2f_dxdy
    
    return [[d2f_dxdx, d2f_dxdy],
            [d2f_dydx, d2f_dydy]]


# Calcul des valeurs propres de la matrice hessienne
points_critiques = [(0, 0), (2, 1)]

for point in points_critiques:

    x, y = point

    matrice = d2f2(x, y)
    valProp = np.linalg.eigvals(matrice)

    print("\nPoint critique :", point)
    print("Matrice hessienne :", matrice)
    print("Valeurs propres :", valProp)

    # Classification du point critique
    if valProp[0] > 0 and valProp[1] > 0:
        print("=> Minimum local")

    elif valProp[0] < 0 and valProp[1] < 0:
        print("=> Maximum local")

    elif valProp[0] * valProp[1] < 0:
        print("=> Point selle")

    else:
        print("=> Test de la Hessienne non concluant")

```

Les résultats obtenus avec la matrice hessienne sont cohérents avec ceux du TD. Le point critique \((0,0)\) est un point selle, tandis que le point \((2,1)\) correspond à un minimum local.
