## Exercice 01


```python
import numpy as np
import matplotlib.pyplot as plt

# Solution
y_s = 7 / 8
x_s = np.sqrt(7 / 32)

# Valeurs
x = np.linspace(-1.2, 1.2, 500)
y = np.linspace(0, 2, 500)

# Fonctions après élimination
f_x = x**2 + (4*x**2 - 1)**2
f_y = y/4 + (y - 1)**2

# Figure
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

# -------------------------
# 1. Méthode de Lagrange
# -------------------------
ax[0].plot(x, 4*x**2, label="y = 4x²")
ax[0].plot(0, 1, "ro", label="P(0,1)")

# Solutions
ax[0].plot(x_s, y_s, "go", label="Solutions")
ax[0].plot(-x_s, y_s, "go")

# Distance minimale
ax[0].plot([0, x_s], [1, y_s], "k--")

ax[0].set_title("Méthode de Lagrange")
ax[0].set_xlabel("x")
ax[0].set_ylabel("y")
ax[0].grid()
ax[0].legend()
ax[0].axis("equal")

# -------------------------
# 2. Élimination de y
# -------------------------
ax[1].plot(x, f_x, label="f(x)")
ax[1].plot([x_s, -x_s],
           [f_x[np.argmin(abs(x-x_s))]] * 2,
           "go", label="Minimum")

ax[1].set_title("Élimination de y")
ax[1].set_xlabel("x")
ax[1].set_ylabel("Distance²")
ax[1].grid()
ax[1].legend()

# -------------------------
# 3. Élimination de x
# -------------------------
ax[2].plot(y, f_y, label="g(y)")
ax[2].plot(y_s, y_s/4 + (y_s-1)**2,
           "go", label="Minimum")

ax[2].set_title("Élimination de x")
ax[2].set_xlabel("y")
ax[2].set_ylabel("Distance²")
ax[2].grid()
ax[2].legend()

plt.tight_layout()
plt.show()
```

## exerice 02

```python
import numpy as np
import matplotlib.pyplot as plt

# Point P
P = np.array([1, 2, 3])

# Solution
PS = P / np.linalg.norm(P)

# Sphère unité
u = np.linspace(0, 2*np.pi, 100)
v = np.linspace(0, np.pi, 100)

X = np.outer(np.cos(u), np.sin(v))
Y = np.outer(np.sin(u), np.sin(v))
Z = np.outer(np.ones_like(u), np.cos(v))

# Distance de chaque point de la sphère à P
D = np.sqrt((X-P[0])**2 + (Y-P[1])**2 + (Z-P[2])**2)

# Graphique
fig = plt.figure(figsize=(8, 7))
ax = fig.add_subplot(111, projection="3d")

# Sphère colorée selon la distance
surface = ax.plot_surface(
    X, Y, Z,
    facecolors=plt.cm.viridis_r(
        (D-D.min())/(D.max()-D.min())
    ),
    alpha=0.8
)

# Point P
ax.scatter(*P, color="red", s=80, label="P(1,2,3)")

# Point solution
ax.scatter(*PS, color="orange", s=120,
           edgecolor="black", label="PS")

# Distance minimale
ax.plot(
    [P[0], PS[0]],
    [P[1], PS[1]],
    [P[2], PS[2]],
    "k--",
    label="Distance minimale"
)

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
ax.set_title("Exercice 2 : point de la sphère le plus proche de P")
ax.legend()

plt.show()
```

## Exercice 03

```python
import numpy as np
import matplotlib.pyplot as plt

# Fonction
def J(x, y):
    return (y - x**2)**2

# Points
A = (1, -2)
B = (1, 1)
S = (-1, 0)

# Valeurs de x
x = np.linspace(-3, 3, 500)

# Graphique
plt.figure(figsize=(8, 8))

# J = 0
plt.plot(x, x**2, "k", label="Noir : J = 0")

# J = 1
plt.plot(x, x**2 + 1, "g", label="Vert : J = 1")
plt.plot(x, x**2 - 1, "g--")

# Contrainte
plt.plot(x, -2*x - 2, "m", label="Violet : contrainte C = 0")

# Points
plt.plot(A[0], A[1], "ro", label="Rouge : point A")
plt.annotate("A(1,-2)", A)

plt.plot(B[0], B[1], "ks", label="Noir carré : point B")
plt.annotate("B(1,1)", B)

plt.plot(S[0], S[1], "*", color="orange", markersize=15,
         label="Orange : solution S")
plt.annotate("S(-1,0)", S)

# Axes
plt.axhline(0, color="gray", linewidth=0.5)
plt.axvline(0, color="gray", linewidth=0.5)

plt.xlim(-3, 3)
plt.ylim(-6, 10)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Isovaleurs de J et contrainte")

# Légende
plt.legend()

plt.grid()
plt.show()

```
